```python
import sys, os, torch, yaml
sys.path.append(os.path.abspath('..'))
from src.transformer.model import GPT
from src.transformer.dataset import get_dataloaders
from src.transformer.generate import load_model, generate_text
```

```python

# Load config
with open('../configs/transformer.yaml') as f:
    config = yaml.safe_load(f)
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
# Option 1: fresh untrained model for shape debugging
train_loader, val_loader, vocab_size, stoi, itos = get_dataloaders(config)
model = GPT(
    vocab_size=vocab_size,
    n_embd=config['model']['n_embd'],
    n_head=config['model']['n_head'],
    n_layer=config['model']['n_layer'],
    block_size=config['data']['block_size'],
).to(device)
```

```python
x, y = next(iter(train_loader))
x, y = x.to(device), y.to(device)
logits, loss = model(x, y)
print('logits', logits.shape)   # (B, T, vocab_size)
print('loss', loss.item())
```
